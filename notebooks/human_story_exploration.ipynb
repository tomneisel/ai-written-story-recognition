```python
import pandas as pd
import seaborn as sns
import plotly.express as px
```

```python
human_stories_df = pd.read_csv("../data/story_features/human_stories.csv")
```

```python
sns.scatterplot(data = human_stories_df, x="sentence_length_variance", y="sentence_length_average", hue="origin")
```

```python
fig = px.scatter_3d(
    human_stories_df,
    "sentence_length_average",
    "sentence_length_variance",
    "sentence_length_max",
    color="origin",
    opacity=0.7
)

fig.show()
```

```python
fig = px.scatter_3d(
    human_stories_df,
    "sentence_length_average",
    "sentence_upper_95",
    "sentence_lower_5",
    color="origin",
    opacity=0.7
)

fig.show()
```
