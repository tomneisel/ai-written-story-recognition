```python
import pandas as pd
import seaborn as sns
import plotly.express as px
```

```python
human_stories_df = pd.read_csv("../data/story_features/human_stories.csv")
ai_stories_df = pd.read_csv("../data/story_features/ai_stories.csv")
stories_df = pd.concat([human_stories_df, ai_stories_df])
```

```python
sns.scatterplot(data = stories_df, x="sentence_length_variance", y="sentence_length_max", hue="origin")
```

```python
fig = px.scatter_3d(
    stories_df,
    "sentence_length_average",
    "sentence_length_variance",
    "sentence_length_max",
    color=stories_df["origin"],
    opacity=0.7
)

fig.show()
```

```python
sns.scatterplot(data = stories_df, x="sentence_length_variance", y="sentence_length_max", hue="written_by")
```

```python
fig = px.scatter_3d(
    stories_df,
    "sentence_length_average",
    "sentence_length_variance",
    "sentence_length_max",
    color=stories_df["written_by"],
    opacity=0.7
)

fig.show()
```
