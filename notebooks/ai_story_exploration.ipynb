```python
import pandas as pd
import seaborn as sns
import plotly.express as px
```

```python
ai_stories_df = pd.read_csv("../data/story_features/ai_stories.csv")
```

```python
sns.scatterplot(data = ai_stories_df, x="sentence_length_variance", y="sentence_length_average", hue="origin")
```

```python
fig = px.scatter_3d(
    x=ai_stories_df["sentence_length_average"],
    y=ai_stories_df["sentence_length_variance"],
    z=ai_stories_df["sentence_length_max"],
    opacity=0.7
)

fig.show()
```
